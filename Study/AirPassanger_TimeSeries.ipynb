#Time Series Forecasting -
The goal of this project is to practice forecasting models that rely on timing. A skill/concept that was not fully explored in the classroom setting and looking to practice application.

The **Air Passangers** dataset from Kaggle, is meant to simulate the number of passagers from 1949 to 1960. Will be an easy dataset to visualize trends/model.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Import Air Passangers Dataset

```python
ap_df = pd.read_csv('AirPassengers.csv')
ap_df.head(5)
```

```python
#Want to determine if the dataset needs to be cleaned
ap_df.info()
```

---
# Clean
## Next Steps:
- The 'Month' column is a string and will need to be formatted into datetime
- Once formatted, time series models typically index based on the date

```python
#Want to convert 'Month' column from string to date format
ap_df['Month'] = pd.to_datetime(ap_df['Month'])
ap_df.set_index(['Month'])
ap_df.info()
```

```python
#What does the new column look like?
ap_df.head(5)
```

---
# Visualizations/ Analysis

```python
plt.title('Airline Passangers From 1950-1969')
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.plot(ap_df['Month'], ap_df['#Passengers'])
plt.show()
```

# Analysis of Visualization 1 -
From year-over-year there is obviously seasonality in the amount of when passangers tend to travel. Not only that, but there seems to be a positive trend as time passes.
